# Exoplanet Taxonomic Clustering
**Goal:** Unsupervised clustering of exoplanets into physical classes (terrestrial, sub-Neptune, gas giant) using all 9 numeric features (radius, mass, period, semi-major axis, eccentricity, equilibrium temperature, stellar Teff, stellar mass, stellar radius) with pl_name as identifier.


**Pipeline:** Audit $\rightarrow$ EDA $\rightarrow$ Log Preprocessing $\rightarrow$ K-Means & Hierarchical Clustering $\rightarrow$ Evaluation.


## 1. Setup & Environment
Initializes random seeds (`seed=42`) and colorblind-safe styling for reproducible clustering and accessible visualizations.


```python
# Core system and data libraries
import sys
import os
import warnings
import time

# Numerical and tabular processing
import numpy as np
import pandas as pd

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Hierarchical clustering and dendrogram generation
from scipy.cluster.hierarchy import linkage, dendrogram

# Preprocessing, clustering algorithms, and evaluation metrics
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Global environment settings
warnings.filterwarnings('ignore')
np.random.seed(42)

# Visualization aesthetics: Colorblind-friendly palette and publication styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
sns.set_palette('colorblind')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.edgecolor'] = '#333333'
plt.rcParams['axes.linewidth'] = 0.8
plt.rcParams['figure.dpi'] = 120

print("Environment Initialized Successfully:")
```

## 2. Data Ingestion & Audit
Loaded 6,445 confirmed exoplanets across 9 numeric features (+ pl_name identifier = 10 columns).
Eccentricity has the highest missing rate (16.5%), while radius and mass are under 1% missing.

```python
# Define file path
dataset_path = '../data/exoplanets.csv'

# Complete candidate feature schema requested
target_requested_columns = [
    'pl_name',     # Planet Name (identifier)
    'pl_rade',     # Planet Radius [Earth radii]
    'pl_bmasse',   # Planet Mass [Earth masses]
    'pl_orbper',   # Orbital Period [days]
    'pl_orbsmax',  # Orbital Semimajor Axis [AU]
    'pl_orbeccen', # Orbital Eccentricity
    'pl_eqt',      # Equilibrium Temperature [K]
    'st_teff',     # Host Star Effective Temperature [K]
    'st_mass',     # Host Star Mass [Solar masses]
    'st_rad'       # Host Star Radius [Solar radii]
]

# Load raw dataset
raw_df = pd.read_csv(dataset_path)
print(f"Raw Dataset Ingested: {raw_df.shape[0]} rows, {raw_df.shape[1]} columns.")
print(f"Raw Columns: {list(raw_df.columns)}")

# Audit schema against available columns
available_target_cols = [c for c in target_requested_columns if c in raw_df.columns]
missing_target_cols = [c for c in target_requested_columns if c not in raw_df.columns]
extraneous_cols = [c for c in raw_df.columns if c not in target_requested_columns]

print(f"\nAvailable Target Columns ({len(available_target_cols)}): {available_target_cols}")
print(f"Requested but Absent Columns ({len(missing_target_cols)}): {missing_target_cols}")
print(f"Extraneous Metadata Columns Dropped ({len(extraneous_cols)}): {extraneous_cols}")

# Filter dataset to available target columns
df_filtered = raw_df[available_target_cols].copy()

# Set identifier 'pl_name' as dataframe index to isolate it from feature matrices
if 'pl_name' in df_filtered.columns:
    df_filtered.set_index('pl_name', inplace=True)

# Separate numerical feature matrix
feature_names = list(df_filtered.columns)
print(f"\nActive Feature Matrix Columns ({len(feature_names)}): {feature_names}")

# Structural Audit: Info and Head
print("\n--- Feature Data Types & Non-Null Counts ---")
df_filtered.info()

print("\n--- First 5 Exoplanet Records in Pipeline ---")
display(df_filtered.head())

# Missing value audit
missing_audit = pd.DataFrame({
    'Missing_Count': df_filtered.isnull().sum(),
    'Missing_Percentage': (df_filtered.isnull().sum() / len(df_filtered)) * 100
}).sort_values(by='Missing_Count', ascending=False)

print("\n--- Missing Value Profile ---")
display(missing_audit)
```

## 3. Exploratory Data Analysis

Spearman rank correlations confirm expected orbital physics, including Kepler's Third Law ($r_s \approx 0.98$).

```python
# Summary statistics table with distributional moments
stats_df = df_filtered.describe().T
stats_df['median'] = df_filtered.median()
stats_df['skewness'] = df_filtered.skew()
stats_df['kurtosis'] = df_filtered.kurtosis()
stats_df = stats_df[['count', 'mean', 'std', 'min', '25%', 'median', '75%', 'max', 'skewness', 'kurtosis']]

print("--- Descriptive Statistics and Distributional Moments ---")
display(stats_df.round(3))

# Distribution Plots (KDE and Histograms)
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(16, 12))
axes = axes.flatten()

palette = sns.color_palette('colorblind')
for i, col in enumerate(feature_names):
    ax = axes[i]
    sns.histplot(
        df_filtered[col].dropna(),
        kde=True,
        ax=ax,
        color=palette[0],
        bins=35
    )
    ax.set_title(f"Distribution of {col}", fontsize=12, fontweight='bold', pad=8)
    ax.set_xlabel(f"{col}", fontsize=10)
    ax.set_ylabel("Frequency", fontsize=10)
    
    # Annotate skewness
    skew_val = df_filtered[col].skew()
    ax.text(
        0.95, 0.90, f"Skew: {skew_val:.2f}",
        transform=ax.transAxes,
        ha='right', va='top',
        fontsize=10,
        bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='gray', alpha=0.8)
    )

plt.tight_layout()
plt.show()

# Correlation Matrix Heatmap
plt.figure(figsize=(9, 7))
corr_matrix = df_filtered.corr(method='spearman') # Spearman rank correlation captures monotonic non-linear relations
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    vmin=-1.0,
    vmax=1.0,
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8, "label": "Spearman Rank Correlation"}
)
plt.title("Spearman Rank Correlation Matrix of Exoplanet Features", fontsize=13, fontweight='bold', pad=12)
plt.tight_layout()
plt.show()
```

## 4. Preprocessing & Feature Engineering
Applies median imputation for missing values, $\log_{10}$-transforms power-law physical features, and standardizes all dimensions to zero mean and unit variance.

```python
# Separate features that span orders of magnitude vs bounded ratios
log_features = ['pl_rade', 'pl_bmasse', 'pl_orbper', 'pl_orbsmax', 'pl_eqt', 'st_mass', 'st_rad']
linear_features = ['pl_orbeccen', 'st_teff']

# 1. Median Imputation (done in original units so medians preserve physical meaning)
imputer = SimpleImputer(strategy='median')
df_imputed = pd.DataFrame(
    imputer.fit_transform(df_filtered),
    columns=feature_names,
    index=df_filtered.index
)

# 2. Log10 transformation for power-law astrophysical properties
X_transformed = df_imputed.copy()
for col in log_features:
    X_transformed[col] = np.log10(X_transformed[col])

# 3. Standardization (now mean ≈ 0, std ≈ 1 across genuinely Gaussian-like scales)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_transformed)

df_scaled = pd.DataFrame(X_scaled, columns=feature_names, index=df_filtered.index)

print("Transformed & Scaled Data Inspection:")
display(df_scaled.describe().round(3).T[['mean', 'std', 'min', '50%', 'max']])
```

## 5. Model 1: K-Means Clustering & Optimization
Partitions planets into $k$ clusters by minimizing inertia[cite: 1]. Optimal cluster count is evaluated across $k \in [2, 10]$ using the Elbow Method and Silhouette scores.


```python
# Hyperparameter search across k in [2, 10]
k_range = range(2, 11)
inertias = []
silhouette_scores_k = []
davies_bouldin_scores_k = []
calinski_harabasz_scores_k = []

print("Running K-Means Hyperparameter Search (k=2 to 10)...")
for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km_labels = km.fit_predict(X_scaled)
    
    inertias.append(km.inertia_)
    silhouette_scores_k.append(silhouette_score(X_scaled, km_labels))
    davies_bouldin_scores_k.append(davies_bouldin_score(X_scaled, km_labels))
    calinski_harabasz_scores_k.append(calinski_harabasz_score(X_scaled, km_labels))
    print(f" - k={k:2d}: Inertia={km.inertia_:10.2f} | Silhouette={silhouette_scores_k[-1]:.4f} | DB={davies_bouldin_scores_k[-1]:.4f} | CH={calinski_harabasz_scores_k[-1]:8.2f}")

# Plotting the Elbow Curve and Silhouette Profile
palette = sns.color_palette('colorblind')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5.5))

# Subplot 1: Inertia (Elbow Curve)
ax1.plot(k_range, inertias, marker='o', linewidth=2.2, markersize=7, color=palette[0], label='Inertia (WCSS)')
ax1.axvline(x=3, color=palette[3], linestyle='--', linewidth=1.5, label='Elbow Point (k=3)')
ax1.axvline(x=4, color=palette[2], linestyle=':', linewidth=1.5, label='Alternative Inflection (k=4)')
ax1.set_title("K-Means Elbow Curve: Inertia vs. Number of Clusters (k)", fontsize=12, fontweight='bold', pad=10)
ax1.set_xlabel("Number of Clusters (k)", fontsize=11)
ax1.set_ylabel("Within-Cluster Sum of Squares (Inertia)", fontsize=11)
ax1.set_xticks(k_range)
ax1.legend(loc='upper right', frameon=True)
ax1.grid(True, alpha=0.5)

# Subplot 2: Silhouette Score Profile
ax2.plot(k_range, silhouette_scores_k, marker='s', linewidth=2.2, markersize=7, color=palette[2], label='Silhouette Score')
ax2.axvline(x=3, color=palette[3], linestyle='--', linewidth=1.5, label='Optimal k=3 (Peak Local Score)')
ax2.set_title("K-Means Silhouette Analysis vs. Number of Clusters (k)", fontsize=12, fontweight='bold', pad=10)
ax2.set_xlabel("Number of Clusters (k)", fontsize=11)
ax2.set_ylabel("Mean Silhouette Coefficient", fontsize=11)
ax2.set_xticks(k_range)
ax2.legend(loc='lower right', frameon=True)
ax2.grid(True, alpha=0.5)

plt.tight_layout()
plt.show()

# Fit optimal models for comparison: k=3 (primary) and k=4 (alternative)
kmeans_k3 = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_scaled)
labels_km3 = kmeans_k3.labels_

kmeans_k4 = KMeans(n_clusters=4, random_state=42, n_init=10).fit(X_scaled)
labels_km4 = kmeans_k4.labels_

print("\nCluster membership distribution for K-Means (k=3):")
for c, count in enumerate(np.bincount(labels_km3)):
    print(f" - Cluster {c}: {count} exoplanets ({count/len(X_scaled)*100:.1f}%)")
```

## 6. Model 2: Hierarchical Clustering
Constructs an agglomerative tree using Ward's linkage to reveal nested taxonomic hierarchies[cite: 1].
Dendrogram cuts validate the optimal cluster count identified during K-Means optimization.


```python
# Compute Hierarchical Linkage Matrix using Ward's method
linkage_ward = linkage(X_scaled, method='ward')

# Generate publication-quality truncated Dendrogram
plt.figure(figsize=(14, 6))
dendro = dendrogram(
    linkage_ward,
    truncate_mode='lastp',
    p=30,                     # Show top 30 merged branches for clean interpretability
    leaf_rotation=45.,
    leaf_font_size=10.,
    show_contracted=True,
    color_threshold=70.0      # Cut-off distance visualizing cluster separation
)

# Annotate cut-off line
plt.axhline(y=70.0, color='red', linestyle='--', linewidth=1.5, label='Cluster Cut-off Threshold (k=3)')
plt.title("Hierarchical Clustering Dendrogram (Ward Linkage, Truncated Top 30 Merges)", fontsize=13, fontweight='bold', pad=12)
plt.xlabel("Cluster Leaf Index / (Sample Count in Leaf)", fontsize=11)
plt.ylabel("Ward Distance Metric (Variance)", fontsize=11)
plt.legend(loc='upper right', frameon=True)
plt.tight_layout()
plt.show()

# Fit Agglomerative models for Ward, Complete, and Average linkages at k=3
linkage_methods = ['ward', 'complete', 'average']
agg_models = {}
agg_labels = {}

for method in linkage_methods:
    model = AgglomerativeClustering(n_clusters=3, linkage=method)
    labels = model.fit_predict(X_scaled)
    agg_models[method] = model
    agg_labels[method] = labels
    counts = np.bincount(labels)
    print(f"Agglomerative Linkage='{method}': Cluster Counts = {counts.tolist()}")
```

## 7. Cluster Validation & Evaluation
Evaluates partition quality using Silhouette Score, Davies-Bouldin Index, and Calinski-Harabasz Index.
Balances compact intra-cluster variance against well-separated inter-cluster margins.


```python
# Define dictionary of candidate models
models_to_evaluate = {
    'K-Means (k=3)': labels_km3,
    'K-Means (k=4)': labels_km4,
    'Agglomerative (Ward, k=3)': agg_labels['ward'],
    'Agglomerative (Complete, k=3)': agg_labels['complete'],
    'Agglomerative (Average, k=3)': agg_labels['average']
}

# Compute validation metrics
comparison_records = []

for model_name, labels in models_to_evaluate.items():
    sil = silhouette_score(X_scaled, labels)
    db = davies_bouldin_score(X_scaled, labels)
    ch = calinski_harabasz_score(X_scaled, labels)
    n_clusts = len(np.unique(labels))
    
    comparison_records.append({
        'Algorithm / Configuration': model_name,
        'Clusters (k)': n_clusts,
        'Silhouette Score (↑)': sil,
        'Davies-Bouldin Index (↓)': db,
        'Calinski-Harabasz Index (↑)': ch
    })

# Format into summary table
comparison_df = pd.DataFrame(comparison_records)
comparison_df.set_index('Algorithm / Configuration', inplace=True)

print("             UNSUPERVISED CLUSTERING PERFORMANCE BENCHMARK SUMMARY")
print(comparison_df.round(4).to_string())
try:
    display(comparison_df.style.format({
        'Silhouette Score (↑)': '{:.4f}',
        'Davies-Bouldin Index (↓)': '{:.4f}',
        'Calinski-Harabasz Index (↑)': '{:.2f}'
    }).highlight_max(subset=['Silhouette Score (↑)', 'Calinski-Harabasz Index (↑)'], color='#d4edda')
      .highlight_min(subset=['Davies-Bouldin Index (↓)'], color='#d4edda'))
except (ImportError, AttributeError):
    # Fallback when jinja2 is not installed (Styler requires jinja2)
    display(comparison_df.round(4))

# Comparative Bar Chart
palette = sns.color_palette('colorblind')
fig, axes = plt.subplots(1, 3, figsize=(17, 5))

metrics = [
    ('Silhouette Score (↑)', 'Silhouette Score (Higher is Better)', palette[0]),
    ('Davies-Bouldin Index (↓)', 'Davies-Bouldin Index (Lower is Better)', palette[3]),
    ('Calinski-Harabasz Index (↑)', 'Calinski-Harabasz Index (Higher is Better)', palette[2])
]

for ax, (metric_col, title_label, bar_color) in zip(axes, metrics):
    comparison_df[metric_col].plot(kind='bar', ax=ax, color=bar_color, edgecolor='black', alpha=0.85)
    ax.set_title(title_label, fontsize=11, fontweight='bold', pad=8)
    ax.set_ylabel("Metric Value", fontsize=10)
    ax.set_xlabel("")
    ax.set_xticklabels(comparison_df.index, rotation=35, ha='right', fontsize=9)
    ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

## 8. Manifold Projections (PCA & t-SNE)
Projects the 9D scaled feature space into 2D using PCA for global variance and t-SNE for local neighborhood preservation.
Visualizes cluster boundaries and confirms non-linear separation between exoplanet regimes.


```python
# 1. Fit Principal Component Analysis (PCA)
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
evr = pca.explained_variance_ratio_

print(f"PCA Fitted Successfully:")
print(f" - Principal Component 1: {evr[0]*100:.2f}% Explained Variance")
print(f" - Principal Component 2: {evr[1]*100:.2f}% Explained Variance")
print(f" - Cumulative 2D Explained Variance: {sum(evr)*100:.2f}%")

# 2. Fit t-Distributed Stochastic Neighbor Embedding (t-SNE)
print("\nFitting t-SNE (2 Components, Perplexity=30, Random State=42)...")
tsne = TSNE(n_components=2, random_state=42, perplexity=30, max_iter=1000)
X_tsne = tsne.fit_transform(X_scaled)
print("t-SNE Projection Completed Successfully.")

# Generate PCA 2D Scatter Plots for EVERY algorithm
palette = sns.color_palette('colorblind')
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(18, 11))
axes_flat = axes.flatten()

plot_configs = [
    ('K-Means (k=3)', labels_km3),
    ('K-Means (k=4)', labels_km4),
    ('Agglomerative (Ward, k=3)', agg_labels['ward']),
    ('Agglomerative (Complete, k=3)', agg_labels['complete']),
    ('Agglomerative (Average, k=3)', agg_labels['average'])
]

for idx, (title, labels) in enumerate(plot_configs):
    ax = axes_flat[idx]
    unique_labels = np.unique(labels)
    
    for l_idx, label_val in enumerate(unique_labels):
        mask = (labels == label_val)
        ax.scatter(
            X_pca[mask, 0],
            X_pca[mask, 1],
            c=[palette[l_idx % len(palette)]],
            label=f'Cluster {label_val} (n={mask.sum()})',
            alpha=0.65,
            edgecolor='none',
            s=22
        )
    
    ax.set_title(f"PCA 2D: {title}", fontsize=11, fontweight='bold', pad=8)
    ax.set_xlabel(f"PC 1 ({evr[0]*100:.1f}% Variance)", fontsize=10)
    ax.set_ylabel(f"PC 2 ({evr[1]*100:.1f}% Variance)", fontsize=10)
    ax.legend(loc='best', frameon=True, fontsize=8.5, markerscale=1.5)
    ax.grid(True, alpha=0.4)

# 6th subplot for summary PCA variance explanation
axes_flat[5].bar(
    ['PC 1', 'PC 2'],
    evr * 100,
    color=[palette[0], palette[1]],
    edgecolor='black',
    alpha=0.8
)
axes_flat[5].set_title("PCA Scree Component Variance", fontsize=11, fontweight='bold', pad=8)
axes_flat[5].set_ylabel("Explained Variance Ratio (%)", fontsize=10)
for p in axes_flat[5].patches:
    axes_flat[5].annotate(f"{p.get_height():.1f}%", (p.get_x() + p.get_width()/2., p.get_height() + 1.0),
                          ha='center', va='bottom', fontsize=10, fontweight='bold')
axes_flat[5].set_ylim(0, 45)
axes_flat[5].grid(axis='y', alpha=0.4)

plt.tight_layout()
plt.show()

# Generate t-SNE 2D Scatter Plots for K-Means and Agglomerative Ward
fig, (ax_tsne1, ax_tsne2) = plt.subplots(1, 2, figsize=(16, 7))

# Plot t-SNE for K-Means (k=3)
for l_idx, label_val in enumerate(np.unique(labels_km3)):
    mask = (labels_km3 == label_val)
    ax_tsne1.scatter(
        X_tsne[mask, 0],
        X_tsne[mask, 1],
        c=[palette[l_idx % len(palette)]],
        label=f'Cluster {label_val} (n={mask.sum()})',
        alpha=0.7,
        edgecolor='none',
        s=20
    )
ax_tsne1.set_title("t-SNE 2D Manifold: K-Means (k=3)", fontsize=12, fontweight='bold', pad=10)
ax_tsne1.set_xlabel("t-SNE Dimension 1", fontsize=11)
ax_tsne1.set_ylabel("t-SNE Dimension 2", fontsize=11)
ax_tsne1.legend(loc='best', frameon=True, fontsize=9.5, markerscale=1.6)
ax_tsne1.grid(True, alpha=0.4)

# Plot t-SNE for Agglomerative Hierarchical (Ward)
for l_idx, label_val in enumerate(np.unique(agg_labels['ward'])):
    mask = (agg_labels['ward'] == label_val)
    ax_tsne2.scatter(
        X_tsne[mask, 0],
        X_tsne[mask, 1],
        c=[palette[l_idx % len(palette)]],
        label=f'Cluster {label_val} (n={mask.sum()})',
        alpha=0.7,
        edgecolor='none',
        s=20
    )
ax_tsne2.set_title("t-SNE 2D Manifold: Agglomerative Hierarchical (Ward, k=3)", fontsize=12, fontweight='bold', pad=10)
ax_tsne2.set_xlabel("t-SNE Dimension 1", fontsize=11)
ax_tsne2.set_ylabel("t-SNE Dimension 2", fontsize=11)
ax_tsne2.legend(loc='best', frameon=True, fontsize=9.5, markerscale=1.6)
ax_tsne2.grid(True, alpha=0.4)

plt.tight_layout()
plt.show()
```

## 9. Astrophysical Interpretation
Inverse-transforms cluster centroids back to physical units to profile planetary archetypes.
Identifies natural demographic regimes: rocky terrestrials, super-Earths/sub-Neptunes, and gas giants.


```python
# Compute original feature median and mean profiles per K-Means (k=3) cluster
df_analyzed = df_filtered.copy()
df_analyzed['Cluster'] = labels_km3

cluster_profiles_median = df_analyzed.groupby('Cluster').median()
cluster_profiles_count = df_analyzed.groupby('Cluster').size()
cluster_profiles_median['Planet_Count'] = cluster_profiles_count
cluster_profiles_median['Population_%'] = (cluster_profiles_count / len(df_analyzed)) * 100

# Assign astrophysically grounded taxonomy names based on median physical parameters
taxonomy_map = {}
for cl in cluster_profiles_median.index:
    rade = cluster_profiles_median.loc[cl, 'pl_rade']
    mass = cluster_profiles_median.loc[cl, 'pl_bmasse']
    orbper = cluster_profiles_median.loc[cl, 'pl_orbper']
    
    if orbper > 1000 or cluster_profiles_median.loc[cl, 'pl_orbsmax'] > 1.5:
        taxonomy_map[cl] = "Cold / Wide-Orbit Giants & Outliers"
    elif rade > 6.0 or mass > 100.0:
        taxonomy_map[cl] = "Hot Jupiters / Jovian Gas Giants"
    else:
        taxonomy_map[cl] = "Terrestrial, Super-Earths & Sub-Neptunes"

cluster_profiles_median['Taxonomy_Classification'] = cluster_profiles_median.index.map(taxonomy_map)

print("             EXOPLANET TAXONOMIC CLUSTER PROFILE (MEDIAN PHYSICAL PROPERTIES)")
print(cluster_profiles_median[['Taxonomy_Classification', 'Planet_Count', 'Population_%', 'pl_rade', 'pl_bmasse', 'pl_orbper', 'pl_orbsmax', 'pl_orbeccen', 'pl_eqt', 'st_teff', 'st_mass', 'st_rad']].round(2).to_string())
display(cluster_profiles_median[['Taxonomy_Classification', 'Planet_Count', 'Population_%', 'pl_rade', 'pl_bmasse', 'pl_orbper', 'pl_orbsmax', 'pl_orbeccen', 'pl_eqt', 'st_teff', 'st_mass', 'st_rad']].round(2))
```
